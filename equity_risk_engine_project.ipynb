```python
"""Indian equity portfolio analytics and quantitative risk assessment.

Run this module to fetch data, analyze individual stocks, evaluate portfolio
construction, and produce a consolidated risk assessment with plots.
Made by Ayush Agarwal
"""

import random
import re

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import yfinance as yf
from arch import arch_model
from scipy.optimize import minimize
from scipy.stats import genpareto
from sklearn.metrics import mean_squared_error, r2_score

TICKERS = [
    "RELIANCE.NS",
    "TCS.NS",
    "HDFCBANK.NS",
    "INFY.NS",
    "ITC.NS",
    "SUNPHARMA.NS",
]
RISK_FREE_RATE = 0.06
TRADING_DAYS = 252

tickers = TICKERS


def make_portfolio(n):
    """Create a random long-only portfolio from the configured ticker universe.

Args:
    n: Number of stocks to select.

Returns:
    DataFrame with Stock and normalized Weight columns."""
    if n < 1 or n > len(tickers):
        raise ValueError(f"Choose n between 1 and {len(tickers)}")

    stocks = random.sample(tickers, n)

    weights = np.random.random(n)
    weights = weights / weights.sum()

    portfolio = pd.DataFrame({
        "Stock": stocks,
        "Weight": weights
    })

    return portfolio


def download_prices(portfolio, period="2y"):
    """Download adjusted daily closing prices for the selected portfolio.

Args:
    portfolio: Holdings DataFrame with a Stock column.
    period: yfinance lookback period.

Returns:
    Date-indexed price DataFrame."""
    stocks = portfolio["Stock"].tolist()

    prices = yf.download(
        tickers=stocks,
        period=period,
        auto_adjust=True,
        progress=False
    )["Close"]

    if isinstance(prices, pd.Series):
        prices = prices.to_frame(name=stocks[0])

    return prices


def fetch_market_returns(period="2y"):
    """Fetch market-index prices and calculate daily market returns.

Args:
    period: yfinance lookback period.

Returns:
    Daily market return Series/DataFrame in the format expected by downstream models."""
    print("---------------")
    print("FETCHING NIFTY 50 DATA")

    benchmark = yf.download(
        "^NSEI",
        period=period,
        auto_adjust=True,
        progress=False
    )["Close"]

    if isinstance(benchmark, pd.DataFrame):
        benchmark = benchmark.iloc[:, 0]

    market_returns = benchmark.pct_change().dropna()
    market_returns.name = "Market"

    print(f"Observations: {len(market_returns)}")
    print(f"Period: {market_returns.index.min().date()} to "
          f"{market_returns.index.max().date()}")

    print("---------------")
    return market_returns


def fetch_indian_fama_french_factors():
    """Fetch and standardize the Indian market, size, value, momentum, and risk-free factors.

Returns:
    Cleaned factor DataFrame indexed by date."""
    url = (
        "https://faculty.iima.ac.in/iffm/"
        "Indian-Fama-French-Momentum/DATA/"
        "2025-12_FourFactors_and_Market_Returns_Daily_"
        "SurvivorshipBiasAdjusted.csv"
    )

    factors = pd.read_csv(url)
    factors.columns = [
        re.sub(r"[^a-z0-9]+", "", str(c).lower())
        for c in factors.columns
    ]

    date_col = next(
        (c for c in factors.columns
         if c in ["date", "dates", "tradingdate"]),
        None
    )

    if date_col is None:
        raise ValueError(
            f"Could not identify date column. "
            f"Downloaded columns: {factors.columns.tolist()}"
        )

    factors[date_col] = pd.to_datetime(
        factors[date_col], errors="coerce"
    )
    factors = factors.dropna(subset=[date_col])
    factors = factors.set_index(date_col).sort_index()

    aliases = {
      "mkt_rf": [
          "mf", "mktrf", "marketpremium",
          "marketfactor", "marketexcessreturn", "mktminusrf"
      ],
      "smb": ["smb", "smallminusbig", "sizefactor"],
      "hml": ["hml", "highminuslow", "valuefactor"],
      "wml": [
          "wml", "mom", "momentum", "momentumfactor",
          "winnersminuslosers"
      ],
      "rf": ["rf", "riskfree", "riskfreerate", "riskfreereturn"],
      "market_return": [
          "marketreturn", "marketindexreturn",
          "market", "bse500return", "niftyreturn"
      ]
  }

    def find_column(key):
        for candidate in aliases[key]:
            if candidate in factors.columns:
                return candidate
        return None

    col_map = {key: find_column(key) for key in aliases}

    if col_map["mkt_rf"] is None:
        raise ValueError(
            "Market premium column not found. "
            f"Downloaded columns: {factors.columns.tolist()}"
        )

    if col_map["rf"] is None:
        if col_map["market_return"] is None:
            raise ValueError(
                "Risk-free rate was not found, and a market-return "
                "column is unavailable to derive it. Inspect the "
                "downloaded CSV headers before proceeding."
            )
        factors["derivedrf"] = (
            pd.to_numeric(factors[col_map["market_return"]], errors="coerce")
            - pd.to_numeric(factors[col_map["mkt_rf"]], errors="coerce")
        )
        col_map["rf"] = "derivedrf"

    required = ["mkt_rf", "smb", "hml", "wml", "rf"]
    missing = [k for k in required if col_map[k] is None]

    if missing:
        raise ValueError(
            f"Missing factor columns: {missing}. "
            f"Downloaded columns: {factors.columns.tolist()}"
        )

    clean = factors[
        [col_map[k] for k in required]
    ].copy()
    clean.columns = ["MKT_RF", "SMB", "HML", "WML", "RF"]

    for col in clean.columns:
        clean[col] = pd.to_numeric(clean[col], errors="coerce")

    clean = clean.dropna()

    print("\nRaw factor data preview:")
    print(clean.head())

    print("\nMaximum absolute value by column:")
    print(clean.abs().max())

    if clean.abs().max().max() > 1:
        clean = clean / 100

    clean = clean[~clean.index.duplicated(keep="last")]

    print("Factor data downloaded successfully.")
    print(f"Date range: {clean.index.min().date()} to "
          f"{clean.index.max().date()}")
    print(f"Observations: {len(clean)}")
    print("Factor columns:", clean.columns.tolist())

    return clean


def calculate_returns(prices, portfolio):
    """Calculate individual-stock and weighted portfolio daily returns.

Args:
    prices: Adjusted close prices by ticker.
    portfolio: Holdings DataFrame with Stock and Weight columns.

Returns:
    Tuple of stock-return DataFrame and portfolio-return Series."""
    stock_returns = prices.pct_change().dropna()

    weights = portfolio.set_index("Stock")["Weight"]
    weights = weights.reindex(stock_returns.columns)

    portfolio_returns = stock_returns.dot(weights)

    return stock_returns, portfolio_returns


def plot_stock_performance(prices):
    """Plot normalized stock price performance and report total returns.

Args:
    prices: Adjusted prices by ticker.

Returns:
    Stock performance table and total-return Series."""
    print("---------------")
    print("INDIVIDUAL STOCK PERFORMANCE")

    prices = prices.dropna(how="all")
    normalized_prices = prices.div(prices.iloc[0]) * 100

    plt.figure(figsize=(13, 6))

    for stock in normalized_prices.columns:
        plt.plot(
            normalized_prices.index,
            normalized_prices[stock],
            label=stock
        )

    plt.axhline(100, color="black", linestyle="--", linewidth=1)
    plt.title("Individual Stock Performance (Starting Value = 100)")
    plt.xlabel("Date")
    plt.ylabel("Normalized Price")
    plt.legend()
    plt.grid(alpha=0.25)
    plt.tight_layout()
    plt.show()

    total_returns = (prices.iloc[-1] / prices.iloc[0] - 1) * 100

    print("\nTOTAL STOCK RETURNS")
    print(total_returns.sort_values(ascending=False).round(2).to_string() + " %")
    print("---------------")

    return normalized_prices, total_returns


def calculate_beta(stock_returns, market_returns, window=126):
    """Estimate full-sample and rolling stock betas against market returns.

Args:
    stock_returns: Daily returns by stock.
    market_returns: Daily market returns.
    window: Rolling beta window in trading days.

Returns:
    Beta results table and rolling-beta DataFrame."""
    print("---------------")
    print("CALCULATING CAPM BETA")

    aligned = stock_returns.join(market_returns, how="inner").dropna()

    if len(aligned) <= window:
        raise ValueError("Insufficient overlapping data for rolling beta.")

    results = []
    rolling_betas = pd.DataFrame(index=aligned.index)

    for stock in stock_returns.columns:

        beta = aligned[stock].cov(aligned["Market"]) / aligned["Market"].var()

        daily_rf = (1 + 0.06) ** (1 / 252) - 1
        stock_excess = aligned[stock] - daily_rf
        market_excess = aligned["Market"] - daily_rf

        from scipy.stats import linregress
        regression = linregress(market_excess, stock_excess)

        results.append({
            "Stock": stock,
            "Beta": beta,
            "Annualized Alpha": regression.intercept * 252,
            "R-squared": regression.rvalue ** 2
        })

        rolling_betas[stock] = (
            aligned[stock].rolling(window).cov(aligned["Market"])
            / aligned["Market"].rolling(window).var()
        )

    results_table = pd.DataFrame(results).set_index("Stock")

    print("\nCAPM RESULTS")
    print(results_table.round(4))
    print("---------------")

    return results_table, rolling_betas


def plot_beta(rolling_betas):
    """Plot rolling stock betas relative to the market.

Args:
    rolling_betas: Rolling beta values by stock."""
    print("---------------")
    print("PLOTTING ROLLING BETA")

    plt.figure(figsize=(13, 6))

    for stock in rolling_betas.columns:
        plt.plot(
            rolling_betas.index,
            rolling_betas[stock],
            label=stock
        )

    plt.axhline(
        1, color="black", linestyle="--",
        label="Beta = 1 (market sensitivity)"
    )
    plt.axhline(0, color="grey", linewidth=0.8)

    plt.title("126-Day Rolling Beta vs NIFTY 50")
    plt.xlabel("Date")
    plt.ylabel("Beta")
    plt.legend()
    plt.grid(alpha=0.25)
    plt.tight_layout()
    plt.show()

    print("---------------")


def run_capm_regression(stock_returns, market_returns, risk_free_rate=0.06):
    """Run stock-level CAPM regressions against market excess returns.

Args:
    stock_returns: Daily returns by stock.
    market_returns: Daily market returns.
    risk_free_rate: Annual risk-free rate as a decimal.

Returns:
    Regression summary and fitted CAPM data as defined by the implementation."""
    from scipy.stats import linregress

    print("---------------")
    print("CAPM REGRESSION")

    aligned = stock_returns.join(market_returns, how="inner").dropna()

    daily_rf = (1 + risk_free_rate) ** (1 / 252) - 1
    market_excess = aligned["Market"] - daily_rf

    results = []

    for stock in stock_returns.columns:
        stock_excess = aligned[stock] - daily_rf

        reg = linregress(market_excess, stock_excess)

        results.append({
            "Stock": stock,
            "Beta": reg.slope,
            "Daily Alpha (%)": reg.intercept * 100,
            "Annualized Alpha (%)": reg.intercept * 252 * 100,
            "R-squared": reg.rvalue ** 2
        })

    results = pd.DataFrame(results).set_index("Stock")

    print(f"Risk-free rate: {risk_free_rate:.2%}")
    print(f"Observations: {len(aligned)}")
    print("\nResults:")
    print(results.round(4))

    print("---------------")
    return results, aligned


def plot_capm_regression(capm_data):
    """Plot the CAPM regression diagnostics returned by run_capm_regression.

Args:
    capm_data: CAPM regression data object."""
    print("---------------")
    print("CAPM REGRESSION PLOTS")

    market = capm_data["Market"]

    for stock in capm_data.columns.drop("Market"):
        x = market.values
        y = capm_data[stock].values

        beta, intercept = np.polyfit(x, y, 1)
        fitted = intercept + beta * x

        plt.figure(figsize=(8, 5))

        plt.scatter(x * 100, y * 100, alpha=0.4, s=14,
                    label="Observed daily returns")

        order = np.argsort(x)
        plt.plot(x[order] * 100, fitted[order] * 100,
                 linewidth=2, label=f"Fitted line (slope = {beta:.2f})")

        plt.xlabel("NIFTY 50 Daily Return (%)")
        plt.ylabel(f"{stock} Daily Return (%)")
        plt.title(f"CAPM Regression: {stock} vs NIFTY 50")
        plt.legend()
        plt.grid(alpha=0.25)
        plt.tight_layout()
        plt.show()

    print("---------------")


def walk_forward_capm(stock_returns, market_returns,
                      train_months=6, test_months=3,
                      risk_free_rate=0.06):
    """Evaluate CAPM coefficients and predictions across sequential train/test windows.

Args:
    stock_returns: Daily returns by stock.
    market_returns: Daily market returns.
    train_months: Training-window length in months.
    test_months: Test-window length in months.
    risk_free_rate: Annual risk-free rate as a decimal.

Returns:
    Predictions, out-of-sample summary, and segment-level results."""

    from scipy.stats import linregress

    print("---------------")
    print("WALK-FORWARD CAPM BACKTEST")

    data = stock_returns.join(market_returns, how="inner").dropna()
    data = data.sort_index()

    daily_rf = (1 + risk_free_rate) ** (1 / 252) - 1
    predictions = {}
    segment_results = []

    start = data.index.min()
    final_date = data.index.max()

    while True:
        train_end = start + pd.DateOffset(months=train_months)
        test_end = train_end + pd.DateOffset(months=test_months)

        train = data[(data.index >= start) & (data.index < train_end)]
        test = data[(data.index >= train_end) & (data.index < test_end)]

        if len(test) == 0 or len(train) < 40:
            break

        for stock in stock_returns.columns:

            x_train = train["Market"] - daily_rf
            y_train = train[stock] - daily_rf

            reg = linregress(x_train, y_train)
            alpha, beta = reg.intercept, reg.slope

            x_test = test["Market"] - daily_rf
            predicted = daily_rf + alpha + beta * x_test
            actual = test[stock]

            if stock not in predictions:
                predictions[stock] = []

            for date in test.index:
                predictions[stock].append({
                    "Date": date,
                    "Actual Return": actual.loc[date],
                    "Predicted Return": predicted.loc[date],
                    "Beta": beta,
                    "Alpha": alpha
                })

            segment_results.append({
                "Stock": stock,
                "Train Start": train.index.min().date(),
                "Train End": train.index.max().date(),
                "Test Start": test.index.min().date(),
                "Test End": test.index.max().date(),
                "Beta": beta,
                "Annualized Alpha (%)": alpha * 252 * 100,
                "Test Observations": len(test),
                "RMSE (%)": np.sqrt(
                    np.mean((actual - predicted) ** 2)
                ) * 100
            })

        start = start + pd.DateOffset(months=test_months)

        if start >= final_date:
            break

    all_predictions = {}
    summary = []

    for stock, records in predictions.items():
        result = pd.DataFrame(records).set_index("Date")
        result = result[~result.index.duplicated(keep="first")]
        all_predictions[stock] = result

        actual = result["Actual Return"]
        predicted = result["Predicted Return"]

        denominator = ((actual - actual.mean()) ** 2).sum()
        oos_r2 = (
            1 - ((actual - predicted) ** 2).sum() / denominator
            if denominator > 0 else np.nan
        )

        summary.append({
            "Stock": stock,
            "OOS R-squared": oos_r2,
            "Actual Mean Daily Return (%)": actual.mean() * 100,
            "Predicted Mean Daily Return (%)": predicted.mean() * 100,
            "RMSE (%)": np.sqrt(np.mean((actual - predicted) ** 2)) * 100
        })

        actual_growth = 100 * (1 + actual).cumprod()
        predicted_growth = 100 * (1 + predicted).cumprod()

        plt.figure(figsize=(12, 5))
        plt.plot(actual_growth.index, actual_growth,
                 label="Actual stock performance", linewidth=2)
        plt.plot(predicted_growth.index, predicted_growth,
                 label="CAPM predicted performance", linewidth=2)

        plt.title(f"{stock}: Walk-Forward CAPM")
        plt.xlabel("Date")
        plt.ylabel("Growth of ₹100")
        plt.legend()
        plt.grid(alpha=0.25)
        plt.tight_layout()
        plt.show()

    summary = pd.DataFrame(summary).set_index("Stock")
    segments = pd.DataFrame(segment_results)

    print(f"Training window: {train_months} months")
    print(f"Testing window: {test_months} months")
    print("\nOUT-OF-SAMPLE RESULTS")
    print(summary.round(4))

    print("\nBETA BY TRAINING/TEST SEGMENT")
    print(segments.round(4).to_string(index=False))

    print("---------------")

    return all_predictions, summary, segments


def run_fama_french_walk_forward(
    stock_returns,
    prices,
    train_months=6,
    test_months=3,
    factors=None
):
    """Fit a four-factor model per stock and evaluate on a later test window.

Args:
    stock_returns: Daily stock returns as decimals.
    prices: Adjusted stock prices used for performance plots.
    train_months: Training-window length in months.
    test_months: Test-window length in months.

Returns:
    Factor coefficients, metrics, predictions, and factor data."""
    if factors is None:
        factors = fetch_indian_fama_french_factors()

    stock_returns = stock_returns.copy()
    prices = prices.copy()

    stock_returns.index = pd.to_datetime(stock_returns.index)
    prices.index = pd.to_datetime(prices.index)

    stock_returns = stock_returns.sort_index()
    prices = prices.sort_index()

    train_days = int(train_months * 21)
    test_days = int(test_months * 21)

    coefficient_rows = []
    metric_rows = []
    prediction_results = {}

    for stock in stock_returns.columns:

        data = pd.concat(
            [
                stock_returns[stock].rename("stock_return"),
                factors
            ],
            axis=1,
            join="inner"
        ).dropna()

        if len(data) < train_days + test_days:
            print(f"Skipping {stock}: insufficient aligned data.")
            continue

        test_start = len(data) - test_days
        train_start = test_start - train_days

        train = data.iloc[train_start:test_start]
        test = data.iloc[test_start:]

        factor_cols = ["MKT_RF", "SMB", "HML", "WML"]

        X_train = np.column_stack([
            np.ones(len(train)),
            train[factor_cols].to_numpy()
        ])
        X_test = np.column_stack([
            np.ones(len(test)),
            test[factor_cols].to_numpy()
        ])

        y_train = (train["stock_return"] - train["RF"]).to_numpy()
        y_test = (test["stock_return"] - test["RF"]).to_numpy()

        beta = np.linalg.lstsq(X_train, y_train, rcond=None)[0]

        fitted_train = X_train @ beta
        fitted_test_excess = X_test @ beta

        predicted_returns = fitted_test_excess + test["RF"].to_numpy()
        actual_returns = test["stock_return"].to_numpy()

        train_r2 = r2_score(y_train, fitted_train)
        test_r2 = r2_score(y_test, fitted_test_excess)
        test_rmse = np.sqrt(
            mean_squared_error(y_test, fitted_test_excess)
        )

        coefficient_rows.append({
            "Stock": stock,
            "Alpha_daily": beta[0],
            "Alpha_annualized_approx": beta[0] * 252,
            "Market_beta": beta[1],
            "SMB_loading": beta[2],
            "HML_loading": beta[3],
            "WML_loading": beta[4],
            "Train_R2": train_r2
        })

        metric_rows.append({
            "Stock": stock,
            "Train_start": train.index.min().date(),
            "Test_start": test.index.min().date(),
            "Test_end": test.index.max().date(),
            "Train_observations": len(train),
            "Test_observations": len(test),
            "Test_R2": test_r2,
            "Test_RMSE_daily": test_rmse
        })

        actual_index = 100 * np.cumprod(1 + actual_returns)
        predicted_index = 100 * np.cumprod(1 + predicted_returns)

        prediction_results[stock] = pd.DataFrame({
            "Actual_Return": actual_returns,
            "Predicted_Return": predicted_returns,
            "Actual_Index": actual_index,
            "Predicted_Index": predicted_index
        }, index=test.index)

    coefficients = pd.DataFrame(coefficient_rows).set_index("Stock")
    metrics = pd.DataFrame(metric_rows).set_index("Stock")

    if coefficients.empty:
        raise ValueError(
            "No stocks could be modelled. Check date overlap "
            "and available price history."
        )

    print("\nFACTOR LOADINGS")
    print("---------------")
    print(coefficients.round(4))

    print("\nOUT-OF-SAMPLE PERFORMANCE")
    print("-------------------------")
    print(metrics.round(4))

    for stock, result in prediction_results.items():
        plt.figure(figsize=(11, 5))
        plt.plot(
            result.index, result["Actual_Index"],
            label="Actual stock performance", linewidth=2
        )
        plt.plot(
            result.index, result["Predicted_Index"],
            label="Fama-French fitted performance",
            linestyle="--", linewidth=2
        )
        plt.axhline(100, linestyle=":", alpha=0.6)
        plt.title(f"{stock}: Actual vs Fama-French fitted performance")
        plt.xlabel("Date")
        plt.ylabel("Growth of 100 (price-return index)")
        plt.legend()
        plt.grid(alpha=0.25)
        plt.tight_layout()
        plt.show()

    print("\nINTERPRETATION GUIDE")
    print("--------------------")
    print("Market_beta > 1: stronger sensitivity to market movements.")
    print("SMB_loading > 0: returns associated with the small-cap factor.")
    print("HML_loading > 0: returns associated with the value factor.")
    print("WML_loading > 0: returns associated with momentum.")
    print("Higher Train_R2: factors explain more training-period variation.")
    print("Test_R2 can be negative when the model performs poorly out of sample.")
    print("Loadings show association, not proof of causation.")

    return coefficients, metrics, prediction_results, factors


def plot_portfolio(portfolio_returns, initial_investment=100000):
    """Plot portfolio value growth and daily returns; print headline performance metrics.

Args:
    portfolio_returns: Daily portfolio returns as decimals.
    initial_investment: Starting portfolio value in currency units.

Returns:
    Portfolio value Series."""
    import matplotlib.pyplot as plt

    portfolio_value = (1 + portfolio_returns).cumprod() * initial_investment

    fig, ax = plt.subplots(2, 1, figsize=(14, 9), sharex=True)

    ax[0].plot(portfolio_value.index, portfolio_value)
    ax[0].set_title("Portfolio Growth")
    ax[0].set_ylabel("Portfolio Value (₹)")
    ax[0].grid(True, alpha=0.3)

    ax[1].plot(portfolio_returns.index, portfolio_returns * 100)
    ax[1].axhline(0, linewidth=1)
    ax[1].set_title("Daily Portfolio Returns")
    ax[1].set_ylabel("Return (%)")
    ax[1].set_xlabel("Date")
    ax[1].grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()

    total_return = (portfolio_value.iloc[-1] / initial_investment - 1) * 100
    annualized_volatility = portfolio_returns.std() * (252 ** 0.5) * 100

    print(f"Initial investment: ₹{initial_investment:,.2f}")
    print(f"Final portfolio value: ₹{portfolio_value.iloc[-1]:,.2f}")
    print(f"Total return: {total_return:.2f}%")
    print(f"Annualized volatility: {annualized_volatility:.2f}%")

    return portfolio_value


def analyze_return_tail(portfolio_returns, confidence_level=0.95):
    """Estimate historical Value at Risk (VaR) and Expected Shortfall at a confidence level.

Args:
    portfolio_returns: Daily portfolio returns as decimals.
    confidence_level: Tail confidence level, such as 0.95 or 0.99.

Returns:
    Dictionary of historical tail-risk metrics."""
    import matplotlib.pyplot as plt
    import numpy as np

    returns = portfolio_returns.dropna()

    if returns.empty:
        raise ValueError("Portfolio returns contain no valid observations.")
    if not 0 < confidence_level < 1:
        raise ValueError("Confidence level must be between 0 and 1.")

    tail_probability = 1 - confidence_level
    var_return = returns.quantile(tail_probability)
    tail_returns = returns[returns <= var_return]
    cvar_return = tail_returns.mean()

    var_loss = -var_return
    cvar_loss = -cvar_return

    mean_return = returns.mean()
    daily_volatility = returns.std()

    fig, ax = plt.subplots(figsize=(12, 6))

    ax.hist(
        returns * 100,
        bins=50,
        density=True,
        alpha=0.65,
        label="Historical daily returns"
    )

    if daily_volatility > 0:
        x = np.linspace(returns.min(), returns.max(), 400)
        normal_curve = (
            np.exp(-0.5 * ((x - mean_return) / daily_volatility) ** 2)
            / (daily_volatility * np.sqrt(2 * np.pi))
        )
        ax.plot(
            x * 100,
            normal_curve / 100,
            linewidth=2,
            label="Normal distribution reference"
        )

    ax.axvspan(
        returns.min() * 100,
        var_return * 100,
        alpha=0.2,
        label=f"Worst {(1 - confidence_level) * 100:.0f}% tail"
    )
    ax.axvline(
        var_return * 100,
        linestyle="--",
        linewidth=2,
        label=f"{confidence_level:.0%} VaR threshold: {var_return * 100:.2f}%"
    )
    ax.axvline(
        cvar_return * 100,
        linestyle=":",
        linewidth=2,
        label=f"{confidence_level:.0%} CVaR: {cvar_return * 100:.2f}%"
    )

    ax.set_title("Portfolio Daily Returns — Historical VaR & CVaR")
    ax.set_xlabel("Daily return (%)")
    ax.set_ylabel("Probability density")
    ax.legend()
    ax.grid(True, alpha=0.25)
    plt.tight_layout()
    plt.show()

    print(f"Confidence level: {confidence_level:.0%}")
    print(f"Observations: {len(returns)}")
    print(f"VaR:  {var_loss * 100:.2f}% daily loss")
    print(f"CVaR: {cvar_loss * 100:.2f}% average tail loss")
    print(f"VaR on ₹1,00,000: ₹{var_loss * 100000:,.2f}")
    print(f"CVaR on ₹1,00,000: ₹{cvar_loss * 100000:,.2f}")

    return {
        "VaR": var_loss,
        "CVaR": cvar_loss,
        "VaR_return_threshold": var_return,
        "CVaR_average_tail_return": cvar_return
    }


def plot_rolling_volatility(portfolio_returns, window=30):
    """Plot rolling and full-sample annualized portfolio volatility.

Args:
    portfolio_returns: Daily portfolio returns as decimals.
    window: Rolling window in trading days.

Returns:
    Rolling annualized volatility Series."""
    import matplotlib.pyplot as plt

    returns = portfolio_returns.dropna()

    if window < 2 or window > len(returns):
        raise ValueError("Window must be between 2 and the number of observations.")

    rolling_vol = returns.rolling(window).std() * (252 ** 0.5) * 100
    overall_vol = returns.std() * (252 ** 0.5) * 100

    plt.figure(figsize=(14, 6))
    plt.plot(rolling_vol.index, rolling_vol, label=f"{window}-day rolling volatility")
    plt.axhline(
        overall_vol,
        linestyle="--",
        linewidth=1.5,
        label=f"Overall annualized volatility: {overall_vol:.2f}%"
    )

    plt.title("Portfolio Rolling Annualized Volatility")
    plt.xlabel("Date")
    plt.ylabel("Annualized volatility (%)")
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()

    print(f"Window: {window} trading days")
    print(f"Overall annualized volatility: {overall_vol:.2f}%")
    print(f"Latest rolling volatility: {rolling_vol.dropna().iloc[-1]:.2f}%")
    print(f"Maximum rolling volatility: {rolling_vol.max():.2f}%")
    print(f"Minimum rolling volatility: {rolling_vol.min():.2f}%")

    return rolling_vol


def fit_garch(portfolio_returns):
    """Fit a GARCH(1,1) model with Student-t innovations to portfolio returns.

Args:
    portfolio_returns: Daily portfolio returns as decimals.

Returns:
    Tuple of fitted ARCH model results and conditional volatility Series."""
    returns = portfolio_returns.dropna() * 100

    model = arch_model(
        returns,
        mean="Constant",
        vol="GARCH",
        p=1,
        q=1,
        dist="t"
    )

    results = model.fit(disp="off")
    conditional_volatility = results.conditional_volatility

    print(results.summary())

    return results, conditional_volatility


def plot_garch_volatility(rolling_volatility, garch_volatility, window=30):
    """Compare realized rolling volatility with GARCH conditional volatility.

Args:
    rolling_volatility: Realized annualized rolling volatility Series.
    garch_volatility: GARCH conditional daily volatility in percentage points.
    window: Smoothing window in trading days."""

    garch_annualized = garch_volatility * np.sqrt(252)

    garch_rolling = (
        np.sqrt(garch_volatility.pow(2).rolling(window).mean())
        * np.sqrt(252)
    )

    fig, ax = plt.subplots(figsize=(14, 6))

    ax.plot(
        rolling_volatility.index,
        rolling_volatility,
        label=f"{window}-day Realized Rolling Volatility",
        alpha=0.85
    )

    ax.plot(
        garch_annualized.index,
        garch_annualized,
        label="Daily GARCH Conditional Volatility",
        alpha=0.45
    )

    ax.plot(
        garch_rolling.index,
        garch_rolling,
        label=f"{window}-day Smoothed GARCH Volatility",
        linewidth=2
    )

    ax.set_title("Realized vs GARCH-Estimated Portfolio Volatility")
    ax.set_xlabel("Date")
    ax.set_ylabel("Annualized Volatility (%)")
    ax.legend()
    ax.grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()

    print(f"Latest realized rolling volatility: {rolling_volatility.dropna().iloc[-1]:.2f}%")
    print(f"Latest daily GARCH volatility: {garch_annualized.dropna().iloc[-1]:.2f}%")
    print(f"Latest smoothed GARCH volatility: {garch_rolling.dropna().iloc[-1]:.2f}%")

    return garch_annualized, garch_rolling


def detect_volatility_regimes(rolling_volatility):
    """Classify rolling volatility into regimes using the function’s existing thresholds.

Args:
    rolling_volatility: Rolling volatility Series.

Returns:
    Tuple containing regime labels and a regime summary."""
    volatility = rolling_volatility.dropna()

    low_threshold = volatility.quantile(1 / 3)
    high_threshold = volatility.quantile(2 / 3)

    regimes = pd.Series(index=volatility.index, dtype="object")

    regimes[volatility <= low_threshold] = "Low Volatility"
    regimes[
        (volatility > low_threshold) &
        (volatility <= high_threshold)
    ] = "Normal Volatility"
    regimes[volatility > high_threshold] = "High Volatility"

    regime_summary = (
        pd.DataFrame({
            "Volatility": volatility,
            "Regime": regimes
        })
        .groupby("Regime")["Volatility"]
        .agg(["count", "mean", "min", "max"])
    )

    regime_summary["Percentage of Days"] = (
        regime_summary["count"] / len(volatility) * 100
    )

    print("Volatility thresholds:")
    print(f"Low / normal boundary: {low_threshold:.2f}%")
    print(f"Normal / high boundary: {high_threshold:.2f}%")
    print("\nRegime Summary:")
    print(regime_summary)

    return regimes, regime_summary


def plot_volatility_regimes(rolling_volatility, volatility_regimes):
    """Visualize volatility regimes and their distribution.

Args:
    rolling_volatility: Rolling volatility Series.
    volatility_regimes: Regime labels aligned to the volatility Series."""
    import matplotlib.pyplot as plt

    fig, ax = plt.subplots(figsize=(15, 6))

    colors = {
        "Low Volatility": "green",
        "Normal Volatility": "yellow",
        "High Volatility": "red"
    }

    regimes = volatility_regimes
    start = 0

    for i in range(1, len(regimes) + 1):
        if i == len(regimes) or regimes.iloc[i] != regimes.iloc[start]:
            regime = regimes.iloc[start]
            ax.axvspan(
                regimes.index[start],
                regimes.index[i - 1],
                color=colors[regime],
                alpha=0.22,
                linewidth=0
            )
            start = i

    vol = rolling_volatility.reindex(regimes.index)
    ax.plot(
        vol.index,
        vol,
        color="black",
        linewidth=1.5,
        label="30-day Rolling Volatility"
    )

    ax.set_title("Portfolio Volatility Regimes")
    ax.set_xlabel("Date")
    ax.set_ylabel("Annualized Volatility (%)")
    ax.grid(True, alpha=0.2)
    ax.legend()

    plt.tight_layout()
    plt.show()


def analyze_portfolio_diversification(prices, portfolio):
    """Calculate portfolio covariance, correlation, and diversification statistics.

Args:
    prices: Adjusted prices by ticker.
    portfolio: Holdings DataFrame with Stock and Weight columns.

Returns:
    Correlation matrix, covariance matrix, and portfolio volatility."""
    stock_returns = prices.pct_change().dropna()

    weights = portfolio.set_index("Stock")["Weight"]
    stock_returns = stock_returns[weights.index]
    weights = weights / weights.sum()

    correlation_matrix = stock_returns.corr()
    covariance_matrix = stock_returns.cov() * 252

    portfolio_variance = weights.values @ covariance_matrix.values @ weights.values
    portfolio_volatility = np.sqrt(portfolio_variance)

    individual_volatilities = stock_returns.std() * np.sqrt(252)
    weighted_average_volatility = (weights * individual_volatilities).sum()

    diversification_benefit = weighted_average_volatility - portfolio_volatility

    print("---------------")
    print("PORTFOLIO DIVERSIFICATION ANALYSIS")

    print("\nCORRELATION MATRIX")
    print(correlation_matrix.round(3))

    print("\nANNUALIZED COVARIANCE MATRIX")
    print(covariance_matrix.round(5))

    print(f"\nPortfolio volatility: {portfolio_volatility:.2%}")
    print(f"Weighted average individual stock volatility: {weighted_average_volatility:.2%}")
    print(f"Diversification benefit: {diversification_benefit:.2%}")

    fig, ax = plt.subplots(figsize=(10, 7))
    image = ax.imshow(correlation_matrix, cmap="RdYlGn_r", vmin=-1, vmax=1)

    ax.set_xticks(range(len(correlation_matrix.columns)))
    ax.set_xticklabels(correlation_matrix.columns, rotation=45, ha="right")
    ax.set_yticks(range(len(correlation_matrix.index)))
    ax.set_yticklabels(correlation_matrix.index)

    for i in range(len(correlation_matrix)):
        for j in range(len(correlation_matrix)):
            ax.text(j, i, f"{correlation_matrix.iloc[i, j]:.2f}",
                    ha="center", va="center", color="black")

    ax.set_title("Stock Return Correlation Heatmap")
    fig.colorbar(image, ax=ax, label="Correlation")
    plt.tight_layout()
    plt.show()

    print("---------------")

    return correlation_matrix, covariance_matrix, portfolio_volatility


def plot_all_rolling_correlations(stock_returns, window=60):
    """Plot rolling pairwise correlations among selected stock returns.

Args:
    stock_returns: Daily returns by stock.
    window: Rolling correlation window in trading days.

Returns:
    Dictionary or collection of rolling pairwise correlations as defined by the implementation."""
    from itertools import combinations

    stock_pairs = list(combinations(stock_returns.columns, 2))
    n_pairs = len(stock_pairs)

    if n_pairs == 0:
        print("---------------")
        print("At least two stocks with valid returns are needed.")
        print("---------------")
        return None

    cols = 2
    rows = int(np.ceil(n_pairs / cols))

    fig, axes = plt.subplots(
        rows, cols, figsize=(16, 4 * rows), squeeze=False
    )
    axes = axes.flatten()

    rolling_correlations = {}

    print("---------------")
    print("ROLLING CORRELATION ANALYSIS")
    print(f"Stocks: {len(stock_returns.columns)}")
    print(f"Unique pairs plotted: {n_pairs}")
    print(f"Rolling window: {window} trading days")

    for i, (stock1, stock2) in enumerate(stock_pairs):
        rolling_corr = stock_returns[stock1].rolling(window).corr(
            stock_returns[stock2]
        )
        rolling_correlations[(stock1, stock2)] = rolling_corr

        axes[i].plot(rolling_corr.index, rolling_corr, linewidth=1.2)
        axes[i].axhline(0, color="black", linewidth=0.8)
        axes[i].set_ylim(-1.05, 1.05)
        axes[i].set_title(f"{stock1} vs {stock2}")
        axes[i].set_ylabel("Correlation")
        axes[i].grid(alpha=0.2)

        full_corr = stock_returns[stock1].corr(stock_returns[stock2])
        latest_corr = rolling_corr.dropna()

        print(
            f"{stock1} vs {stock2}: "
            f"Full-period = {full_corr:.3f}, "
            f"Latest rolling = "
            f"{latest_corr.iloc[-1]:.3f}" if not latest_corr.empty
            else f"{stock1} vs {stock2}: insufficient data"
        )

    for j in range(n_pairs, len(axes)):
        axes[j].set_visible(False)

    fig.suptitle(
        f"{window}-Day Rolling Correlations Across All Stock Pairs",
        fontsize=14
    )
    plt.tight_layout()
    plt.show()

    print("---------------")

    return rolling_correlations


def calculate_sharpe_ratio(portfolio_returns, risk_free_rate=0.06):
    """Calculate the annualized Sharpe ratio from daily portfolio returns.

Args:
    portfolio_returns: Daily portfolio returns as decimals.
    risk_free_rate: Annual risk-free rate as a decimal.

Returns:
    Annualized Sharpe ratio."""
    print("---------------")
    print("OVERALL SHARPE RATIO")

    daily_rf = (1 + risk_free_rate) ** (1 / 252) - 1

    excess_returns = portfolio_returns.dropna() - daily_rf

    sharpe = (
        excess_returns.mean() / portfolio_returns.dropna().std()
    ) * np.sqrt(252)

    print(f"Annual Risk-Free Rate: {risk_free_rate:.2%}")
    print(f"Annualized Sharpe Ratio: {sharpe:.4f}")
    print("---------------")

    return sharpe


def plot_rolling_sharpe(portfolio_returns, window=60, risk_free_rate=0.06):
    """Plot rolling annualized Sharpe ratios.

Args:
    portfolio_returns: Daily portfolio returns as decimals.
    window: Rolling window in trading days.
    risk_free_rate: Annual risk-free rate as a decimal.

Returns:
    Rolling Sharpe Series."""
    print("---------------")
    print("ROLLING SHARPE RATIO")

    daily_rf = (1 + risk_free_rate) ** (1 / 252) - 1
    returns = portfolio_returns.dropna()

    excess_returns = returns - daily_rf

    rolling_mean = excess_returns.rolling(window).mean()
    rolling_std = returns.rolling(window).std()

    rolling_sharpe = (rolling_mean / rolling_std) * np.sqrt(252)

    plt.figure(figsize=(14, 5))
    plt.plot(rolling_sharpe.index, rolling_sharpe, linewidth=1.2)
    plt.axhline(0, color="black", linewidth=0.8)
    plt.axhline(
        rolling_sharpe.mean(), color="red",
        linestyle="--", label="Average rolling Sharpe"
    )
    plt.title(f"{window}-Day Rolling Sharpe Ratio")
    plt.xlabel("Date")
    plt.ylabel("Annualized Sharpe Ratio")
    plt.legend()
    plt.grid(alpha=0.25)
    plt.tight_layout()
    plt.show()

    print(f"Rolling window: {window} trading days")
    print(f"Valid rolling observations: {rolling_sharpe.notna().sum()}")
    print("---------------")

    return rolling_sharpe


def optimize_portfolio_weights(stock_returns, portfolio, risk_free_rate=0.06):
    """Optimize portfolio weights and compare candidate portfolio strategies.

Args:
    stock_returns: Daily returns by stock.
    portfolio: Current holdings DataFrame.
    risk_free_rate: Annual risk-free rate as a decimal.

Returns:
    Optimized weights, optimization results, and strategy return Series/DataFrame."""
    from scipy.optimize import minimize

    print("---------------")
    print("PORTFOLIO OPTIMIZATION")

    returns = stock_returns.dropna()
    stocks = returns.columns
    n = len(stocks)

    if n < 1 or len(returns) < 2:
        raise ValueError("Insufficient return data for optimization.")

    current_weights = (
        portfolio.set_index("Stock")["Weight"]
        .reindex(stocks)
        .to_numpy(dtype=float)
    )
    current_weights = current_weights / current_weights.sum()

    expected_returns = returns.mean().to_numpy() * 252
    covariance = returns.cov().to_numpy() * 252

    covariance = (covariance + covariance.T) / 2

    def metrics(weights):
        annual_return = weights @ expected_returns
        variance = weights @ covariance @ weights
        volatility = np.sqrt(max(variance, 0))
        sharpe = (
            (annual_return - risk_free_rate) / volatility
            if volatility > 0 else np.nan
        )
        return annual_return, volatility, sharpe

    def volatility_objective(weights):
        return metrics(weights)[1]

    def negative_sharpe_objective(weights):
        return -metrics(weights)[2] if metrics(weights)[2] == metrics(weights)[2] else 1e10

    constraints = [{"type": "eq", "fun": lambda w: np.sum(w) - 1}]
    bounds = [(0, 1)] * n
    equal_weights = np.ones(n) / n

    min_result = minimize(
        volatility_objective,
        equal_weights,
        method="SLSQP",
        bounds=bounds,
        constraints=constraints,
        options={"maxiter": 2000, "ftol": 1e-12}
    )

    sharpe_result = minimize(
        negative_sharpe_objective,
        equal_weights,
        method="SLSQP",
        bounds=bounds,
        constraints=constraints,
        options={"maxiter": 2000, "ftol": 1e-12}
    )

    if not min_result.success or not sharpe_result.success:
        raise RuntimeError(
            f"Optimization failed: {min_result.message}; "
            f"{sharpe_result.message}"
        )

    strategies = {
        "Current Random": current_weights,
        "Equal Weight": equal_weights,
        "Minimum Volatility": min_result.x,
        "Maximum Sharpe": sharpe_result.x
    }

    weights_table = pd.DataFrame(strategies, index=stocks).T

    results = []
    strategy_returns = {}

    for name, weights in strategies.items():
        portfolio_series = returns @ weights
        strategy_returns[name] = portfolio_series

        annual_return, annual_volatility, sharpe = metrics(weights)

        results.append({
            "Strategy": name,
            "Expected Annual Return": annual_return,
            "Annualized Volatility": annual_volatility,
            "Sharpe Ratio": sharpe
        })

    results_table = pd.DataFrame(results).set_index("Strategy")

    print("\nPORTFOLIO WEIGHTS (%)")
    print((weights_table * 100).round(2))

    print("\nESTIMATED PERFORMANCE")
    print(results_table.round(4))

    print("\nAssumptions: long-only, fully invested,")
    print(f"annual risk-free rate = {risk_free_rate:.1%}")
    print("Estimates use historical average returns and covariance.")
    print("---------------")

    return weights_table, results_table, strategy_returns


def plot_portfolio_sharpe_comparison(
    strategy_returns, risk_free_rate=0.06, window=60
):
    """Compare overall and rolling Sharpe ratios across portfolio strategies.

Args:
    strategy_returns: Returns for the candidate strategies.
    risk_free_rate: Annual risk-free rate as a decimal.
    window: Rolling window in trading days.

Returns:
    Overall Sharpe ratios and rolling Sharpe DataFrame."""
    print("---------------")
    print("SHARPE RATIO COMPARISON")

    daily_rf = (1 + risk_free_rate) ** (1 / 252) - 1

    overall_sharpes = {}
    rolling_sharpes = {}

    for name, series in strategy_returns.items():
        series = series.dropna()
        excess = series - daily_rf

        overall_sharpes[name] = (
            excess.mean() / series.std() * np.sqrt(252)
            if series.std() > 0 else np.nan
        )

        rolling_mean = excess.rolling(window).mean()
        rolling_std = series.rolling(window).std()

        rolling_sharpes[name] = (
            rolling_mean / rolling_std
        ) * np.sqrt(252)

    overall_sharpes = pd.Series(overall_sharpes).sort_values()

    plt.figure(figsize=(11, 5))
    overall_sharpes.plot(kind="bar")
    plt.axhline(0, color="black", linewidth=0.8)
    plt.title("Overall Annualized Sharpe Ratio by Strategy")
    plt.ylabel("Sharpe Ratio")
    plt.xlabel("")
    plt.xticks(rotation=20, ha="right")
    plt.grid(axis="y", alpha=0.25)
    plt.tight_layout()
    plt.show()

    plt.figure(figsize=(14, 6))

    for name, series in rolling_sharpes.items():
        plt.plot(series.index, series, label=name, linewidth=1.4)

    plt.axhline(0, color="black", linewidth=0.8)
    plt.title(f"{window}-Day Rolling Sharpe Ratio: Portfolio Comparison")
    plt.ylabel("Annualized Sharpe Ratio")
    plt.xlabel("Date")
    plt.legend()
    plt.grid(alpha=0.25)
    plt.tight_layout()
    plt.show()

    print("\nOVERALL SHARPE RATIOS")
    print(overall_sharpes.sort_values(ascending=False).round(4))
    print("---------------")

    return overall_sharpes, pd.DataFrame(rolling_sharpes)


def walk_forward_portfolio_backtest(
    stock_returns, portfolio, risk_free_rate=0.06, window_months=6
):
    """Evaluate portfolio optimization in sequential train/test windows to reduce look-ahead bias.

Args:
    stock_returns: Daily returns by stock.
    portfolio: Current holdings DataFrame.
    risk_free_rate: Annual risk-free rate as a decimal.
    window_months: Training window length in months.

Returns:
    Backtest result object(s) produced by the implementation."""
    from scipy.optimize import minimize
    from pandas.tseries.offsets import DateOffset

    print("---------------")
    print("WALK-FORWARD PORTFOLIO BACKTEST")

    returns = stock_returns.dropna().sort_index()
    stocks = returns.columns
    n = len(stocks)

    random_weights = (
        portfolio.set_index("Stock")["Weight"]
        .reindex(stocks)
        .to_numpy(dtype=float)
    )
    random_weights /= random_weights.sum()

    equal_weights = np.ones(n) / n
    daily_rf = (1 + risk_free_rate) ** (1 / 252) - 1

    strategies = [
        "Original Random",
        "Equal Weight",
        "Minimum Volatility",
        "Maximum Sharpe"
    ]

    test_returns = {name: [] for name in strategies}
    period_results = []

    train_start = returns.index[0]
    final_date = returns.index[-1]

    while True:
        train_end = train_start + DateOffset(months=window_months)
        test_end = train_end + DateOffset(months=window_months)

        if test_end > final_date:
            break

        train = returns.loc[
            (returns.index >= train_start) & (returns.index < train_end)
        ]
        test = returns.loc[
            (returns.index >= train_end) & (returns.index < test_end)
        ]

        if len(train) < 60 or len(test) == 0:
            train_start = train_end
            continue

        mu = train.mean().to_numpy() * 252
        cov = train.cov().to_numpy() * 252
        cov = (cov + cov.T) / 2

        def volatility(w):
            return np.sqrt(max(w @ cov @ w, 0))

        def negative_sharpe(w):
            vol = volatility(w)
            if vol <= 0:
                return 1e10
            return -((w @ mu - risk_free_rate) / vol)

        constraints = [{"type": "eq", "fun": lambda w: np.sum(w) - 1}]
        bounds = [(0, 1)] * n
        initial = np.ones(n) / n

        min_result = minimize(
            volatility, initial, method="SLSQP",
            bounds=bounds, constraints=constraints
        )
        sharpe_result = minimize(
            negative_sharpe, initial, method="SLSQP",
            bounds=bounds, constraints=constraints
        )

        if not min_result.success or not sharpe_result.success:
            print(f"Optimization failed for training window starting {train_start.date()}.")
            train_start = train_end
            continue

        weights_by_strategy = {
            "Original Random": random_weights,
            "Equal Weight": equal_weights,
            "Minimum Volatility": min_result.x,
            "Maximum Sharpe": sharpe_result.x
        }

        for name, weights in weights_by_strategy.items():
            series = test @ weights
            test_returns[name].append(series)

            std = series.std()
            sharpe = (
                ((series.mean() - daily_rf) / std) * np.sqrt(252)
                if std > 0 else np.nan
            )

            period_results.append({
                "Train Start": train.index[0].date(),
                "Train End": train.index[-1].date(),
                "Test Start": test.index[0].date(),
                "Test End": test.index[-1].date(),
                "Strategy": name,
                "Test Sharpe": sharpe,
                "Test Return": (1 + series).prod() - 1,
                "Test Volatility": std * np.sqrt(252),
                "Weights": np.round(weights, 4).tolist()
            })

        train_start = train_end

    combined_returns = {
        name: pd.concat(chunks).sort_index()
        for name, chunks in test_returns.items() if chunks
    }

    if not combined_returns:
        print("Not enough data for a complete train/test window.")
        print("---------------")
        return None

    summary = []

    for name, series in combined_returns.items():
        std = series.std()
        sharpe = (
            ((series.mean() - daily_rf) / std) * np.sqrt(252)
            if std > 0 else np.nan
        )

        summary.append({
            "Strategy": name,
            "Total Test Return": (1 + series).prod() - 1,
            "Annualized Return": series.mean() * 252,
            "Annualized Volatility": std * np.sqrt(252),
            "Sharpe Ratio": sharpe
        })

    summary_table = pd.DataFrame(summary).set_index("Strategy")
    periods_table = pd.DataFrame(period_results)

    print("\nOUT-OF-SAMPLE SUMMARY")
    print(summary_table.round(4))

    print("\nPERIOD-BY-PERIOD RESULTS")
    print(
        periods_table[
            ["Test Start", "Test End", "Strategy",
             "Test Sharpe", "Test Return", "Test Volatility"]
        ].round(4).to_string(index=False)
    )

    plt.figure(figsize=(13, 6))

    for name, series in combined_returns.items():
        growth = (1 + series).cumprod()
        plt.plot(growth.index, growth, label=name)

    plt.axhline(1, color="black", linewidth=0.8)
    plt.title("Walk-Forward Out-of-Sample Portfolio Growth")
    plt.xlabel("Date")
    plt.ylabel("Growth of 1 Unit Invested")
    plt.legend()
    plt.grid(alpha=0.25)
    plt.tight_layout()
    plt.show()

    plt.figure(figsize=(13, 6))

    for name, series in combined_returns.items():
        rolling_mean = (series - daily_rf).rolling(60).mean()
        rolling_std = series.rolling(60).std()
        rolling_sharpe = rolling_mean / rolling_std * np.sqrt(252)
        plt.plot(rolling_sharpe.index, rolling_sharpe, label=name)

    plt.axhline(0, color="black", linewidth=0.8)
    plt.title("60-Day Rolling Sharpe — Walk-Forward Test Periods")
    plt.xlabel("Date")
    plt.ylabel("Annualized Sharpe Ratio")
    plt.legend()
    plt.grid(alpha=0.25)
    plt.tight_layout()
    plt.show()

    print("---------------")

    return summary_table, periods_table, combined_returns


def analyze_concentration_risk(portfolio):
    """Measure stock- and industry-level exposure concentration using HHI.

Args:
    portfolio: Holdings DataFrame with Stock and Weight columns.

Returns:
    Dictionary containing holdings, industry exposure, HHI values, and effective counts."""
    print("---------------")
    print("PORTFOLIO CONCENTRATION RISK")
    print("---------------")

    industry_map = {
        "RELIANCE.NS": "Energy & Diversified",
        "TCS.NS": "IT Services",
        "HDFCBANK.NS": "Banking",
        "INFY.NS": "IT Services",
        "ITC.NS": "Consumer Goods",
        "SUNPHARMA.NS": "Pharmaceuticals"
    }

    holdings = portfolio[["Stock", "Weight"]].copy()

    if holdings.empty:
        raise ValueError("Portfolio cannot be empty.")

    if holdings["Weight"].isna().any() or (holdings["Weight"] < 0).any():
        raise ValueError("Weights must be non-negative and non-missing.")

    if holdings["Weight"].sum() <= 0:
        raise ValueError("Portfolio weights must sum to a positive value.")

    holdings["Weight"] /= holdings["Weight"].sum()

    unknown = set(holdings["Stock"]) - set(industry_map)
    if unknown:
        raise ValueError(f"Add industry classifications for: {sorted(unknown)}")

    holdings["Industry"] = holdings["Stock"].map(industry_map)

    stock_hhi = (holdings["Weight"] ** 2).sum()
    effective_stocks = 1 / stock_hhi

    industry_exposure = (
        holdings.groupby("Industry")["Weight"]
        .sum()
        .sort_values(ascending=False)
    )

    industry_hhi = (industry_exposure ** 2).sum()
    effective_industries = 1 / industry_hhi

    print("\nSTOCK-LEVEL EXPOSURE")
    print(
        holdings.sort_values("Weight", ascending=False)
        .to_string(index=False, formatters={"Weight": "{:.2%}".format})
    )

    print("\nINDUSTRY EXPOSURE")
    print(
        industry_exposure.to_frame("Weight")
        .to_string(formatters={"Weight": "{:.2%}".format})
    )

    print("\nCONCENTRATION METRICS")
    print(f"Stock HHI: {stock_hhi:.4f} ({stock_hhi * 10000:.1f} on 10,000 scale)")
    print(f"Effective number of stocks: {effective_stocks:.2f}")
    print(f"Industry HHI: {industry_hhi:.4f} ({industry_hhi * 10000:.1f} on 10,000 scale)")
    print(f"Effective number of industries: {effective_industries:.2f}")
    print(f"Largest industry: {industry_exposure.index[0]}")
    print(f"Largest industry exposure: {industry_exposure.iloc[0]:.2%}")

    print("---------------")

    return {
        "holdings": holdings,
        "industry_exposure": industry_exposure,
        "stock_hhi": stock_hhi,
        "industry_hhi": industry_hhi,
        "effective_stocks": effective_stocks,
        "effective_industries": effective_industries
    }


def analyze_portfolio_evt(portfolio_returns, threshold_quantile=0.95):
    """Fit a Generalized Pareto distribution to portfolio-loss exceedances and estimate tail VaR/ES.

Args:
    portfolio_returns: Daily portfolio returns as decimals.
    threshold_quantile: Quantile defining the peaks-over-threshold cutoff.

Returns:
    Risk-estimate table and fitted EVT model details."""

    returns = pd.Series(portfolio_returns).dropna().astype(float)
    losses = -returns.to_numpy()

    n = len(losses)
    if n < 100:
        raise ValueError("Too few observations for a meaningful EVT analysis.")

    threshold = np.quantile(losses, threshold_quantile)
    excesses = losses[losses > threshold] - threshold
    n_exceedances = len(excesses)

    if n_exceedances < 15:
        raise ValueError(
            f"Only {n_exceedances} threshold exceedances. "
            "Consider a lower threshold, such as the 90th percentile."
        )

    exceedance_probability = n_exceedances / n

    shape, loc, scale = genpareto.fit(excesses, floc=0)

    if scale <= 0:
        raise ValueError("Invalid fitted GPD scale parameter.")

    print(f"Observations: {n}")
    print(f"Threshold: {threshold:.4%} daily loss")
    print(f"Exceedances: {n_exceedances} ({exceedance_probability:.2%})")
    print(f"GPD shape (xi): {shape:.4f}")
    print(f"GPD scale (beta): {scale:.4%}")

    def evt_var(confidence):
        tail_probability = 1 - confidence

        if tail_probability >= exceedance_probability:
            raise ValueError(
                f"{confidence:.0%} confidence is too low for this "
                "threshold. Choose a lower threshold."
            )

        ratio = exceedance_probability / tail_probability

        if abs(shape) < 1e-6:
            return threshold + scale * np.log(ratio)

        return threshold + (scale / shape) * (ratio**shape - 1)

    def evt_es(confidence):
        var = evt_var(confidence)

        if shape >= 1:
            return np.nan

        return (var + scale - shape * threshold) / (1 - shape)

    rows = []

    for confidence in [0.95, 0.99]:
        var_evt = evt_var(confidence)
        es_evt = evt_es(confidence)

        var_hist = np.quantile(losses, confidence)
        es_hist = losses[losses >= var_hist].mean()

        hits = int(np.sum(losses > var_evt))
        expected_hits = n * (1 - confidence)

        rows.append({
            "Confidence": f"{confidence:.0%}",
            "Historical_VaR": var_hist,
            "EVT_VaR": var_evt,
            "Historical_ES": es_hist,
            "EVT_ES": es_evt,
            "Observed_EVT_VaR_hits": hits,
            "Expected_hits_if_calibrated": expected_hits,
            "Observed_hit_rate": hits / n
        })

    results = pd.DataFrame(rows).set_index("Confidence")

    print("\nRISK ESTIMATES (daily loss)")
    print("---------------------------")
    print((results[
        ["Historical_VaR", "EVT_VaR", "Historical_ES", "EVT_ES"]
    ] * 100).round(3).astype(str) + "%")

    print("\nIN-SAMPLE EXCEEDANCE CHECK")
    print("--------------------------")
    print(results[
        ["Observed_EVT_VaR_hits", "Expected_hits_if_calibrated",
         "Observed_hit_rate"]
    ].round(4))

    sorted_excesses = np.sort(excesses)
    empirical_survival = (
        n_exceedances - np.arange(n_exceedances)
    ) / n_exceedances

    x_grid = np.linspace(0, sorted_excesses.max(), 300)
    fitted_survival = genpareto.sf(
        x_grid, c=shape, loc=0, scale=scale
    )

    plt.figure(figsize=(10, 6))
    plt.step(
        sorted_excesses * 100, empirical_survival,
        where="post", label="Empirical exceedance survival"
    )
    plt.plot(
        x_grid * 100, fitted_survival,
        linestyle="--", linewidth=2, label="Fitted GPD survival"
    )
    plt.yscale("log")
    plt.xlabel("Loss excess above threshold (%)")
    plt.ylabel("Probability excess exceeds x (log scale)")
    plt.title("Portfolio Loss Tail: Empirical vs GPD Fit")
    plt.legend()
    plt.grid(alpha=0.25)
    plt.tight_layout()
    plt.show()

    plt.figure(figsize=(10, 5))
    plt.hist(
        excesses * 100, bins="auto", density=True,
        alpha=0.5, label="Observed loss excesses"
    )
    plt.plot(
        x_grid * 100,
        genpareto.pdf(x_grid, c=shape, loc=0, scale=scale) / 100,
        linewidth=2, label="Fitted GPD density"
    )
    plt.xlabel("Loss excess above threshold (%)")
    plt.ylabel("Density")
    plt.title("Generalized Pareto Distribution Fit")
    plt.legend()
    plt.grid(alpha=0.25)
    plt.tight_layout()
    plt.show()

    if shape >= 1:
        print(
            "\nWARNING: Estimated shape >= 1. The fitted GPD has "
            "infinite theoretical Expected Shortfall; EVT_ES is NaN."
        )

    print("\nINTERPRETATION")
    print("--------------")
    print("VaR is a loss threshold; ES is the average loss beyond VaR.")
    print("EVT extrapolates beyond the threshold using the fitted tail.")
    print("Compare both plots: large deviations suggest a poor tail fit.")
    print("The hit counts are in-sample diagnostics, not independent validation.")
    print("Threshold choice and the small number of extreme observations "
          "can materially change the results.")

    return results, {
        "threshold": threshold,
        "shape": shape,
        "scale": scale,
        "excesses": excesses,
        "fitted_gpd": (shape, scale)
    }


# Made by Ayush Agarwal

def main():
    """Run the complete portfolio analysis in a deliberate, readable order."""
    print("\n" + "=" * 72)
    print("INDIAN EQUITY PORTFOLIO RISK ANALYTICS")
    print("=" * 72)

    # 1. Configure the holdings and fetch all primary market data first.
    portfolio = make_portfolio(3)
    print("\nSELECTED PORTFOLIO")
    print(portfolio.to_string(index=False, formatters={"Weight": "{:.2%}".format}))
    print(f"Total portfolio weight: {portfolio['Weight'].sum():.2%}")

    prices = download_prices(portfolio, period="2y")
    if prices.empty:
        raise RuntimeError("Price download returned no data. Check tickers or data access.")
    stock_returns, portfolio_returns = calculate_returns(prices, portfolio)
    market_returns = fetch_market_returns(period="2y")
    factors = fetch_indian_fama_french_factors()

    # 2. Understand each holding before combining the holdings into a portfolio.
    print("\n" + "=" * 72 + "\n1. STOCK-LEVEL PERFORMANCE AND MARKET RISK\n" + "=" * 72)
    stock_performance, stock_total_returns = plot_stock_performance(prices)
    beta_results, rolling_betas = calculate_beta(stock_returns, market_returns, window=126)
    plot_beta(rolling_betas)
    capm_results, capm_data = run_capm_regression(stock_returns, market_returns, risk_free_rate=RISK_FREE_RATE)
    plot_capm_regression(capm_data)
    capm_predictions, capm_oos_summary, capm_segments = walk_forward_capm(
        stock_returns, market_returns, train_months=6, test_months=3,
        risk_free_rate=RISK_FREE_RATE
    )
    ff_coefficients, ff_metrics, ff_predictions, _ = run_fama_french_walk_forward(
        stock_returns=stock_returns, prices=prices, train_months=6, test_months=3,
        factors=factors
    )

    # 3. Show realized portfolio outcome and core risk measures.
    print("\n" + "=" * 72 + "\n2. PORTFOLIO PERFORMANCE AND DIVERSIFICATION\n" + "=" * 72)
    portfolio_value = plot_portfolio(portfolio_returns)
    tail_metrics = analyze_return_tail(portfolio_returns, confidence_level=0.95)
    overall_sharpe = calculate_sharpe_ratio(portfolio_returns, risk_free_rate=RISK_FREE_RATE)
    rolling_sharpe = plot_rolling_sharpe(portfolio_returns, window=60, risk_free_rate=RISK_FREE_RATE)
    rolling_volatility = plot_rolling_volatility(portfolio_returns, window=30)
    garch_results, garch_volatility = fit_garch(portfolio_returns)
    plot_garch_volatility(rolling_volatility, garch_volatility, window=30)
    volatility_regimes, regime_summary = detect_volatility_regimes(rolling_volatility)
    plot_volatility_regimes(rolling_volatility, volatility_regimes)
    correlation_matrix, covariance_matrix, portfolio_volatility = analyze_portfolio_diversification(prices, portfolio)
    rolling_correlations = plot_all_rolling_correlations(stock_returns, window=60)

    # 4. Compare current weights with optimized alternatives and an out-of-sample test.
    print("\n" + "=" * 72 + "\n3. PORTFOLIO WEIGHT ALTERNATIVES\n" + "=" * 72)
    optimized_weights, optimization_results, strategy_returns = optimize_portfolio_weights(
        stock_returns, portfolio, risk_free_rate=RISK_FREE_RATE
    )
    overall_sharpes, rolling_sharpes = plot_portfolio_sharpe_comparison(
        strategy_returns, risk_free_rate=RISK_FREE_RATE
    )
    walk_forward_results = walk_forward_portfolio_backtest(
        stock_returns, portfolio, risk_free_rate=RISK_FREE_RATE, window_months=6
    )

    # 5. Concentration and extreme-loss analysis; interpret these as different risk lenses.
    print("\n" + "=" * 72 + "\n4. CONCENTRATION AND EXTREME-LOSS RISK\n" + "=" * 72)
    concentration_results = analyze_concentration_risk(portfolio)
    evt_results, evt_model = analyze_portfolio_evt(portfolio_returns, threshold_quantile=0.95)

    # 6. Concise end-of-run interpretation, using the calculated results.
    print("\n" + "=" * 72 + "\nPORTFOLIO RISK SUMMARY\n" + "=" * 72)
    print(f"Portfolio total return: {(portfolio_value.iloc[-1] / 100000 - 1):.2%}")
    print(f"Annualized Sharpe ratio: {overall_sharpe:.3f}")
    print(f"Annualized realized volatility: {portfolio_returns.std() * np.sqrt(TRADING_DAYS):.2%}")
    print(f"Stock concentration HHI: {concentration_results['stock_hhi']:.4f}; effective holdings: {concentration_results['effective_stocks']:.2f}")
    print(f"Industry concentration HHI: {concentration_results['industry_hhi']:.4f}; effective industries: {concentration_results['effective_industries']:.2f}")
    print(f"Largest industry exposure: {concentration_results['industry_exposure'].index[0]} ({concentration_results['industry_exposure'].iloc[0]:.2%})")
    if not evt_results.empty:
        print("\nEVT tail-risk estimates (daily loss):")
        print(evt_results.to_string())
    print("\nInterpretation guide:")
    print("- CAPM beta measures market sensitivity; factor loadings describe historical return associations.")
    print("- Sharpe compares excess return with volatility; it does not describe tail losses by itself.")
    print("- HHI measures exposure concentration, not the full risk contribution of correlated holdings.")
    print("- EVT extrapolates losses beyond a chosen threshold; estimates are sensitive to threshold and sample size.")
    print("- Optimized weights and backtests are model-dependent; out-of-sample performance is not guaranteed.")

    return {
        "portfolio": portfolio, "prices": prices, "stock_returns": stock_returns,
        "portfolio_returns": portfolio_returns, "stock_performance": stock_performance,
        "beta_results": beta_results, "capm_results": capm_results,
        "capm_oos_summary": capm_oos_summary, "ff_metrics": ff_metrics,
        "portfolio_value": portfolio_value, "tail_metrics": tail_metrics,
        "overall_sharpe": overall_sharpe, "rolling_volatility": rolling_volatility,
        "garch_results": garch_results, "regime_summary": regime_summary,
        "correlation_matrix": correlation_matrix, "optimized_weights": optimized_weights,
        "optimization_results": optimization_results, "walk_forward_results": walk_forward_results,
        "concentration_results": concentration_results, "evt_results": evt_results,
        "evt_model": evt_model
    }


if __name__ == "__main__":
    results = main()
```
